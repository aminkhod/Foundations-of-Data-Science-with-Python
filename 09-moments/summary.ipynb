# Summary

Expected values, and especially moments, are commonly used to characterize distributions. In this chapter, we first introduced expected value and moments and demonstrated how to calculate them and explained some of their useful properties. Then we discussed parameter estimation, focusing on estimation of the mean and variance of a random distribution based on observed samples from that distribution. We showed how to find confidence intervals using bootstrap resampling and by using analytical methods for the special case of the mean estimator. When we do not know the standard deviation or variance of the distribution, this resulted in needing to introduce a new random distribution called Student's $t$ distribution. We also showed how to conduct an analytical NHST (commonly called a T-test when the underlying variances are not known) for a difference of means between two groups. 

We also introduced the sampling distribution for an estimator, and revisited the concept of a bootstrap distribution in the context of an estimator. These two distributions are closely related, and we showed by example how the bootstrap distribution can converge to the sampling distribution as the number of data samples becomes large ($ \ge 100$). Finally, we introduced the concept of effect size and showed how to calculate the number of data samples needed for a given combination of signficance, power, and effect size.
